```python
from transformers import pipeline
```

```python
# Huggingface login

from huggingface_hub import notebook_login
notebook_login()
```

```python
#model_card = "Qwen/Qwen2.5-7B-Instruct"
m = "meta-llama/Llama-3.2-1B"
chatPipe = pipeline('text-generation', model = m)
```

```python
while True:
    user_input = input("Ask Anything...")
    print("Human: ", user_input)
    if user_input.lower() in ['quit', 'bye', 'exit']:
        break
    response = chatPipe(user_input)[0]['generated_text']

    print("AI Response: ", response)
```
