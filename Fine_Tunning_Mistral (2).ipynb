```python
# Import os
import os

# PyTorch core library
import torch

# load_dataset - Downloads and loads a dataset (from the Hugging Face Hub or local files). Dataset - to create a custom dataset
from datasets import load_dataset, Dataset

# Prepare model for PEFT Training
# LoraConfig = Configuration class for LoRA (Low‑Rank Adaptation). Defines parameters like rank, alpha, dropout, and which layers to apply LoRA to.
# AutoPeftModelForCausalLM = Wrapper that automatically loads a causal language model (like GPT, LLaMA) with PEFT applied. Lets you use quantized
# or adapter‑based fine‑tuned models easily.
# prepare_model_for_kbit_training = Prepares a model for k‑bit quantized training (like 4‑bit or 8‑bit). Ensures layers are compatible with
# quantization and gradient checkpointing.
# get_peft_model = Applies a PEFT method (like LoRA) to your base model. Wraps the original model with adapter layers so you only fine‑tune a
# small fraction of parameters.
from peft import LoraConfig, AutoPeftModelForCausalLM, prepare_model_for_kbit_training, get_peft_model

# AutoModelForCausalLM = Loads a causal language model (like GPT, LLaMA, Mistral) for text generation tasks.
# AutoTokenizer = Loads the tokenizer that matches the model (handles text → tokens and tokens → text).
# GPTQConfig = Configuration for GPTQ quantization (a method to compress models into lower precision like 4‑bit). Defines parameters such as bits,
# group size, and dataset used for calibration.
# TrainingArguments = A Hugging Face class that defines training hyperparameters (batch size, learning rate, epochs, logging, evaluation strategy,
# etc.).
from transformers import AutoModelForCausalLM, AutoTokenizer, GPTQConfig, TrainingArguments

# SFTTrainer = A specialized trainer class for fine‑tuning causal language models (like GPT, LLaMA, Mistral) using supervised datasets.
# SFTConfig = Configuration class for the SFTTrainer. Defines hyperparameters like batch size, learning rate, max sequence length, logging,
# evaluation strategy, etc.
from trl import SFTTrainer, SFTConfig

# Warnings
import warnings
warnings.filterwarnings("ignore")
```

```python
# Load a 4-bit quantized model

from transformers import BitsAndBytesConfig

quantization_config = BitsAndBytesConfig(

    # This enables 4-bit quantization for the model,  it instructs the loader to quantize the model weights to 4 bits instead of the usual
    # 32-bit floating-point representation
    load_in_4bit=True,

    # 4-bit quantization type This specifies the type of 4-bit quantization to use. “nf4” stands for “Normal Float 4”, a quantization method
    # designed to be more suitable for language models compared to standard 4-bit quantization
    bnb_4bit_quant_type="nf4",

    # This sets the data type used for computations during inference or training. “float16” specifies that calculations should be done in
    # half-precision floating-point format.
    bnb_4bit_compute_dtype=torch.float16,

    # Use double quantization for memory efficiency , Enables nested quantization, also known as “double quantization”. When set to True, it
    # applies a second level of quantization to the already quantized weights. The first quantization reduces weights to 4 bits, and the second
    # quantization further compresses the scales and zeros of the first quantization. This can provide additional memory savings with minimal
    # impact on model quality.
    bnb_4bit_use_double_quant=True,
)
```

```python
# Loading the tokenizer of Mistral model 7B

model_name = "mistralai/Mistral-7B-Instruct-v0.1"
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
# Prepares your model for low‑bit (k‑bit) training — typically 4‑bit or 8‑bit quantization.
# Ensures the model’s layers are compatible with quantized weights (e.g., replacing certain modules with quantization‑friendly versions).
# Sets up gradient checkpointing and other tweaks so training remains stable even when precision is reduced.
# Makes it possible to fine‑tune very large models (like LLaMA‑7B, Mistral‑7B) on GPUs with limited VRAM.

model = prepare_model_for_kbit_training(model)
```

```python
model
```

```python
# LORA Configuration
peft_config = LoraConfig(
        r=16,
        lora_alpha=16,
        lora_dropout=0.05,
        bias="none",
        task_type="CAUSAL_LM",
        target_modules=["q_proj", "v_proj"]
    )
```

```python
#quantised Model
# Wraps the base model with adapter layers defined in your peft_config (for example, LoRA).
# Freezes most of the original parameters so they don’t get updated during training.
# Adds small trainable modules (like low‑rank matrices in LoRA) that capture task‑specific knowledge.
# Returns a new model object that looks like the original but is now PEFT‑enabled.

#model = get_peft_model(model, peft_config)
```

```python
# Setting of training hyperparameters

training_arguments = TrainingArguments(
        output_dir="mistral-finetuned-samsum",
        per_device_train_batch_size=8,
        gradient_accumulation_steps=4,
        optim="paged_adamw_32bit",
        learning_rate=2e-4,
        lr_scheduler_type="cosine",
        save_strategy="epoch",
        logging_steps=100,
        num_train_epochs=1,
        max_steps=1000,
        fp16=True,
        push_to_hub = True,
        gradient_checkpointing=True
  )
```

```python
# SFTConfig
# Defines training hyperparameters (batch size, epochs, logging, saving).
# Here:
#    output_dir="./results" → where checkpoints/logs go.
#    per_device_train_batch_size=8 → batch size per GPU.
#    num_train_epochs=1 → one epoch of training.
#    logging_steps=25 → log every 25 steps.
#    save_strategy="no" → don’t save checkpoints during training.

# SFTTrainer - Wraps Hugging Face’s Trainer but optimized for LLMs + PEFT.

# Takes in:
#    model=model → your base model (already wrapped with LoRA adapters).
#    train_dataset=data → the dataset you want to fine‑tune on.
#    args=sft_args → the training config.
#    peft_config=peft_config → tells trainer to apply LoRA/PEFT during training.

from trl import SFTConfig

sft_args = SFTConfig(
    output_dir="./results",
    per_device_train_batch_size=8,
    num_train_epochs=1,
    logging_steps=25,
    save_strategy="no",
)

trainer = SFTTrainer(
    model=model,
    train_dataset=data,
    args=sft_args,
    peft_config=peft_config,
)
```

```python
trainer.train()
```

```python
trainer.save_model()
tokenizer.save_pretrained(sft_args.output_dir)
```
