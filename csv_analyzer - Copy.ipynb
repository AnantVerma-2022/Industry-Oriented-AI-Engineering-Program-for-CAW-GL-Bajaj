#### LangchainCSV Agent


A LangChain CSV Agent allows you to interact with tabular CSV files using natural language. Under the hood, it loads the CSV data into a Pandas DataFrame and provides the language model (LLM) with a Python REPL tool. When you ask a question, the LLM writes Python code, executes it locally via the tool, and parses the output to deliver a human-readable response

```python
from langchain_experimental.agents.agent_toolkits import create_csv_agent
```

```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(temperature=0.5)
```

```python
llm
```

```python
from re import VERBOSE
agent_executer = create_csv_agent(llm,
                                  "E:\\Batch_3_GenAI\\demo_langchain\\titanic.csv",
                                  #verbose=True,
                                  allow_dangerous_code=True)
```

```python
data['Age'].max()
```

```python
agent_executer.invoke("What is the maximum age in Age column?")
```
